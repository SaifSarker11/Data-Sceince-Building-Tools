```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = [1, 2, 3]
y = [4, 5, 6]
a = np.array(x)
b = np.array(y)
```

```python
plt.plot(a, b, marker = "o", color = "#0000FF", linestyle = ":", linewidth = 2)
plt.grid()
plt.xlabel("X Axis")
plt.ylabel("Y Axis")
plt.title("Line Plot")
plt.show()
```

```python
import pandas as pd
```

```python
data = {
    "Salary" : [25000, 30000, 37000, 28000, 39000, 48000, 55000, 52000, 35000, 26000, 27000, 31000, 33000, 29000, 40000, 36000, 34000, 45000, 47000, 50000]
}
```

```python
df = pd.DataFrame(data)
```

```python
plt.plot(df["Salary"], marker = "o", markeredgecolor = "black", color = "#FF0000", linestyle = "--", linewidth = 2)
plt.ylabel("Salary")
plt.title("Line Plot")
plt.show()
```

##**Histogram**

Too few bins (e.g., bins=2) will oversimplify data, hiding important details and patterns. (**Underfitting**)

Too many bins (e.g., bins=200) will make the histogram too noisy, showing gaps and random fluctuations instead of the overall trend. (**Overfitting**)

```python
rgb_color = [112, 234, 220]
teal_color = (rgb_color[0]/255.0, rgb_color[1]/255.0, rgb_color[2]/255.0)
plt.hist(df["Salary"], bins = 10, edgecolor = "#000000", color = teal_color)
plt.xlabel("Salary Range")
plt.ylabel("Frequency")
plt.title("Histogram")
plt.show()
```

##**Univariate (Categorical)**

```python
df["Dept"] = ["HR", "IT", "Finance", "HR", "Finance", "IT", "HR", "Finance", "IT", "HR"] * 2
```

##**Pie Chart**

```python
count = df["Dept"].value_counts()
color_picker = ["crimson", "beige", "brown"]
plt.pie(count, labels = count.index, autopct = "%1.1f%%", colors = color_picker, shadow = True, explode = [0, 0.2, 0.1])
plt.title("Pie Chart depicting Dept. Ratio")
plt.show()
```

##**Count plot**

```python
plt.bar(count.index, count, color = ["blue", "white", "red"], edgecolor = "black")
plt.ylabel("Dept. Count")
plt.show()
```

##**Bivariate (Numerical, Numerical)**

```python
df["Age"] = [22, 25, 29, 24, 30, 35, 40, 36, 28, 22, 23, 25, 26, 24, 31, 29, 27, 33, 35, 38]
```

##**Scatter Plot**

```python
plt.scatter(df["Age"], df["Salary"], color = "orange", s = 50)
plt.xlabel("Age")
plt.ylabel("Salary")
plt.show()
```

##**Line Plot**

```python
sorted_df = df.sort_values("Age")
plt.plot(sorted_df["Age"], sorted_df["Salary"], marker = "o", markerfacecolor = "white", linewidth = 2, color = "cyan")
plt.xlabel("Age")
plt.ylabel("Salary")
plt.show()
```

##**Bar Chart**

```python
plt.bar(df["Age"], df["Salary"], color = "#136207")
plt.xlabel("Age")
plt.ylabel("Salary")
plt.show()
```

##**Bivariate (Numerical, Categorical)**

##**Box Plot**

```python
HR_Salary = df[df["Dept"] == "HR"]["Salary"]
IT_Salary = df[df["Dept"] == "IT"]["Salary"]
Finance_Salary = df[df["Dept"] == "Finance"]["Salary"]
plt.boxplot([HR_Salary, IT_Salary, Finance_Salary], label = ["HR", "IT", "Finance"])
plt.ylabel("Salary")
plt.show()
```

##**Pie Chart**

```python
avg_salary_by_dept = df.groupby("Dept")["Salary"].mean()
plt.pie(avg_salary_by_dept, autopct = "%1.1f", shadow = True, explode = [0.1, 0, 0.1], labels = ["HR", "IT", "Finance"])
plt.axis("equal") #Ensures the pie chart renders as a perfect circle instead of an ellipse
plt.show()
```

##**Bar Plot**

```python
HR_Salary_Mean = sum(HR_Salary)/len(HR_Salary)
IT_Salary_Mean = sum(IT_Salary)/len(IT_Salary)
Finance_Salary_Mean = sum(Finance_Salary)/len(Finance_Salary)
plt.bar(["HR", "IT", "Finance"], [HR_Salary_Mean, IT_Salary_Mean, Finance_Salary_Mean], color = ["Green", "White", "Red"], edgecolor = "Black")
plt.ylabel("Average Salary")
plt.show()
```

##**Multivariate**

```python
df["Experience"] =  [1, 2, 4, 1.5, 5, 8, 12, 10, 4, 1, 1.2, 2, 3, 1.8, 5.5, 4.5, 3.2, 7, 8, 10]
```

##**Scatter Plot**

```python
plt.scatter(df["Age"], df["Salary"], s = df["Experience"]*100, color = "Skyblue", edgecolor = "Black")
plt.xlabel("Age")
plt.ylabel("Salary")
plt.show()
```

```python
Dept_Color_Dict = {"HR" : "Crimson", "IT" : "Beige", "Finance" : "Navy"}
for Department, Color_Choice in Dept_Color_Dict.items():
  df_dept_wise = df[df["Dept"] == Department]
  plt.scatter(df_dept_wise["Age"], df_dept_wise["Salary"], c = Color_Choice, s = 100, label = Department)
plt.xlabel("Age")
plt.ylabel("Salary")
plt.legend()
```

##**Matplotlib using Object Oriented API**

```python
fig, axs = plt.subplots(1, 3, figsize = (20, 5))

#Line Plot
asc_ordered_df = df.sort_values("Age")
axs[0].plot(asc_ordered_df["Age"], asc_ordered_df["Salary"], color = "Firebrick", marker = "o", linewidth = 2)
axs[0].grid()
axs[0].set_title("Line Plot")
axs[0].set_xlabel("Age")
axs[0].set_ylabel("Salary")

#Histogram
axs[1].hist(df["Salary"], bins = 15, color = "cornflowerblue", edgecolor = "Black")
axs[1].set_title("Histogram")
axs[1].set_xlabel("Salary")
axs[1].set_ylabel("Frequency")

#Boxplot:
axs[2].boxplot(df["Salary"])
axs[2].set_title("Boxplot")
axs[2].set_ylabel("Salary")

plt.show()
```

```python
data_table = {
    "Year": [2020, 2021, 2022, 2023],
    "Sales": [100, 150, 200, 250],
    "Profit": [20, 30, 40, 50],
    "Expenses": [80, 120, 160, 200]
}
dataf = pd.DataFrame(data_table)
plt.plot(dataf["Year"], dataf["Sales"], label = "Sales")
plt.plot(dataf["Year"], dataf["Profit"], label = "Profit")
plt.plot(dataf["Year"], dataf["Expenses"], label = "Expenses")
plt.xlabel("Year")
plt.ylabel("Amount")
plt.legend()
plt.show()
```

##**3D Plot**

```python
fig = plt.figure(figsize = (20, 12))
ax = fig.add_subplot(projection = "3d") #Case Sensitive
ax.scatter(df["Age"], df["Salary"], df["Experience"], s = 500)
ax.set_xlabel("Age")
ax.set_ylabel("Salary")
ax.set_zlabel("Experience")
plt.show()
```
